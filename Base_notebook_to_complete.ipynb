## 1. La démarche

Les données sont réparties en quatre dossiers (`public_records`, `health`, `ecology`, `logistics`). On s'est donc réparti les dossiers : l'objectif était de nettoyer et d'explorer les données brutes.

Chacun (ou par binôme) a travaillé sur ses données avec sa méthode, mais pour chaque dossier on a suivi la même logique :
1. **nettoyer** : doublons, espaces, dates, nombres, majuscules (la base du nettoyage de données vue en cours) ;
2. **trouver les incohérences** : valeurs extrêmes, groupes atypiques, textes différents des autres ;
3. **mettre en commun les anomalies** : un nom ou un code trouvé dans une table est recherché dans les autres tables, puis dans les autres dossiers.

Le détail de ces étapes se trouve dans les notebooks de chaque dossier (`notebook-public_record/`, `notebook-health/`, `notebook-ecology/`, `notebook-logistics/`). Une fois les analyses individuelles terminées, nous avons croisé les données des différents dossiers dans le notebook secret.ipynb afin d'identifier des liens entre nos découvertes et de formuler des hypothèses sur le secret recherché.
Ce rapport résume ce qu'on a compris du secret.

## 2. Nettoyage des données

Les problèmes rencontrés étaient similaires d'un dossier à l'autre :
- des doublons (lignes répétées) ;
- des espaces en trop et des majuscules incohérentes (`ACTIVE` / `active`, `vALENNE rEPUBLIC`), qui faisaient compter une même catégorie plusieurs fois ;
- des dates dans plusieurs formats (`2021-06-08`, `19/12/2018`, `19-Nov-2020`) ;
- des nombres écrits comme du texte (`84,9`, `82.2 index`, `96.0 %`) ;
- des colonnes inintéressantes (ex : `channel_count`, `amplitude_unit` ou `archive_status` dans les enregistrements cardiaques, qui ont la même valeur sur toutes les lignes).

On a donc corrigé ces formats et vérifié à chaque fois qu'aucune valeur n'était perdue. On a essayé de ne pas corriger les erreurs de fond mais de les garder en tête, parce qu'elles pouvaient avoir un sens. Par exemple, 9 mandats de dirigeants se terminent avant d'avoir commencé, et environ 50 % des organisations citées dans les journaux d'événements n'existent pas dans le registre officiel.

**Erreur faite au début :** comme les données étaient vraiment nombreuses, on a d'abord essayé de supprimer des colonnes ou des données pour y voir plus clair et alléger ce qu'il y avait à traiter. Mais avec du recul, cette méthode introduit un biais de jugement : on n'est pas sûr que ce qu'on enlève soit vraiment inutile. Par exemple, au départ on avait moyenné les signaux cardiaques par enregistrement, pour avoir une seule valeur au lieu de centaines, mais cela effaçait la forme du signal. On est donc revenu aux signaux bruts.

## 3. Analyse dossier par dossier

### 3.1 Public records : une identité qui traverse les siècles
En triant les licences par date, on a trouvé deux licences de 1886 et 1895 **toujours actives**. L'une appartient à **Emil Warden**, né en 1871. Dans une lettre de 1942, Warden annonce qu'il voyagera « sous les papiers Vosmere » et demande de ne pas mentionner son ancien nom. Dans le registre des dirigeants, les mandats s'enchaînent : **Warden** (1895-1943), **Adrian Vosmere** (1952-1991), puis **Dr. Elias Voss** (depuis 1996). Une note d'archive ajoute que plusieurs de ces noms partagent la même adresse et les mêmes expressions.

En regardant l'heure des accès aux sites naturels, on a repéré 18 visites à **3h du matin**, toutes faites par *Horizon Field Research Cooperative*, souvent avec du matériel de stockage à froid. Un message de lanceur d'alerte (Mara Venn) explique que des chauves-souris *Eonycteris spelaea* sont collectées, gardées au froid, puis retirées de l'inventaire, et que les documents parlent de **« consommation »**. Le message cite le code **CHAMELEO**. Les mémos internes montrent que CHAMELEO est une filière confidentielle contrôlée par la direction de **Meridian BioLogistics**, dont le président est **Elias Voss**. Une autorité nommée **« Unwabu »** y approuve les exceptions.

### 3.2 Ecology : des chauves-souris qui disparaissent
Les relevés montrent des chutes brutales d'*Eonycteris spelaea* sur certaines grottes. Par exemple, sur le site KNT-KES, la population passe de 21 individus à 3 en quelques semaines. Pourtant, les capteurs fonctionnent normalement au même moment (plus de 90 % de disponibilité, aucun code de panne). Les interventions sont déclarées comme de la « maintenance de routine », et un rapport de terrain (EFR-510088) signale du matériel coupé et des traces de passage inhabituelles. Une équipe de prestataires (*eco-service-1*) concentre presque toutes les pannes déclarées (79, contre moins de 5 pour les autres).

La présence des chauves-souris est corrélée (0,47) au taux de nouaison des vergers voisins : leur disparition a donc aussi un effet sur la pollinisation.

### 3.3 Logistics : des expéditions qui disparaissent des stocks
On est parti de l'accès nocturne à KNT-KES (21-22 mars 2024). Six heures après la fin de cet accès, l'expédition `SHP-2024-00007` part d'un dépôt voisin avec 6 lots de « matériel biologique » sous 2-8 °C. Elle va de **Horizon** vers **Meridian BioLogistics**. Huit jours après l'arrivée, un ajustement de stock (`ADJ-2024-00117`) retire 3,02 lots, sans numéro de réception associé.

Sur l'ensemble des données, 16 des 18 lignes d'inventaire liées à une expédition de ce type ont un écart de stock négatif (écart moyen de −7,28, contre +0,01 pour les autres lignes).

### 3.4 Health : des patients qui ne sont pas comme les autres
Dans les notes médicales, on a mesuré pour chaque note la part de phrases qu'on ne retrouve dans aucune autre note. 26 notes sur 30 sont des modèles recopiés, mais **4 notes sont entièrement uniques**. Elles décrivent toutes un tableau étrange : deux sources de battements cardiaques, une démarche qui se balance malgré un très bon équilibre, des yeux qui bougent indépendamment, et un visage que les caméras et les capteurs thermiques n'arrivent pas à capter correctement. Un seul nom propre apparaît dans toutes les notes : **Elias Voss**. Une note précise qu'un dossier patient peut lui être relié en croisant plusieurs informations.

On a ensuite construit une table avec une ligne par patient (la moyenne de chaque type de mesure), puis appliqué un **k-means**. Avec k=3, un petit groupe de **15 patients sur 240** se détache : signaux thoraciques élevés, balancement important et reconnaissance par caméra très faible. Les 4 notes uniques concernent toutes ce groupe.

Enfin, en traçant les ECG bruts, on a vu que chez ces patients les deux canaux ne sont pas synchronisés : l'un bat à environ 65 bpm, l'autre à 85-98 bpm. Cela correspond à **deux cœurs**. C'est le cas pour 8 des 9 enregistrements du groupe, contre 6 sur 35 chez les autres patients.

## 4. Croisement des résultats

En reliant les dossiers, plusieurs choses coïncident :

- **Ecology et logistics** : un délai d'exactement six heures se répète entre la fin des interventions et le départ des expéditions. Sur les 15 cas analysables, l'activité des chauves-souris baisse ensuite de 38 à 77 %, à chaque fois plus que sur les sites témoins de la même région.
- **Logistics et health** : Meridian livre aussi **Asterion Clinical Institute** (26 livraisons), et ces livraisons précèdent de quelques jours des procédures médicales confidentielles appelées **MRP-42** (30 correspondances).
- **Health, Logistics et Public Records :** le témoignage **PUB-510079** indique que des chauves-souris *Eonycteris spelaea* seraient collectées, transportées dans des unités scellées puis retirées du circuit habituel des stocks, **au moins une partie étant destinée à la consommation**. Le témoignage mentionne également le programme **CHAMELEO**, qu'un mémo confidentiel (**MEM-510101**) associe à des autorisations restreintes et à la direction de Meridian.
- **Public records et health** : Elias Voss, président de Meridian, est le nom cité dans la note médicale. En cherchant les patients qui correspondent à son profil public (né vers 1961, région de Valenne, suivi par le service *executive health*), un seul fait aussi partie du groupe atypique et a un ECG à deux rythmes : **PAT-31162, Eli Varen**, né en 1962, avec les mêmes initiales.

## 5. Conclusion : le narratif caché

Le secret qu'on propose est le suivant : **Elias Voss**, président de Meridian BioLogistics, ne serait pas humain. Il ferait partie d'un petit groupe d'individus aux caractéristiques proches de celles d'un **caméléon** (deux cœurs, des yeux indépendants, un balancement, une apparence qui change), ce qui correspond au nom du programme, CHAMELEO. Il vivrait depuis plus d'un siècle en changeant d'identité (Warden, Vosmere, Voss).

Pour rester en vie ou ne pas vieillir, il **consommerait des chauves-souris** *Eonycteris spelaea*. Elles sont capturées la nuit par Horizon sous couvert de recherche scientifique, expédiées au froid vers Meridian, puis effacées des stocks. Ce trafic fait chuter les populations de chauves-souris et, par conséquent, la pollinisation des vergers.
